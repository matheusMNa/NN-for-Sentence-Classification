Tomando decisões relvantes:

```python
import pickle
from pathlib import Path

data_dir = Path('../../data/processed')  # ajuste conforme a localização do novo notebook

with open(data_dir / 'sequences_pubmed_rct20k.pkl', 'rb') as f:
    seq_data = pickle.load(f)

X_train_idx = seq_data['train']
y_train = seq_data['y_train']

print(len(X_train_idx), len(y_train))  # devem bater: 176642 e 176642
```

Descartando sentenças de comprimento 0:

```python
import numpy as np

comprimentos = np.array([len(s) for s in X_train_idx])
idx_vazios = np.where(comprimentos == 0)[0]
print(f"Índices vazios: {idx_vazios.tolist()}")
print(f"Rótulos correspondentes: {[y_train[i] for i in idx_vazios]}")

mask = comprimentos > 0
X_train_idx = [s for s, m in zip(X_train_idx, mask) if m]
y_train = [y for y, m in zip(y_train, mask) if m]
print(f"Novo tamanho do treino: {len(X_train_idx)}")  # esperado: 176639
```
